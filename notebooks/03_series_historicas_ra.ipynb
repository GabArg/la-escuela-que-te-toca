# Series históricas RA 2011–2025
EDA longitudinal de conteos con comparabilidad auditada. Las clasificaciones son reglas exploratorias, no scores ni recomendaciones.

```python
import pandas as pd
from src.ingestion.ra_historico import write_long, validate_long
from src.analysis.ra_historico import build_indicators, classify_series
```

```python
long = write_long()
validate_long(long)
panel = build_indicators(long)
senales = classify_series(panel)
long.groupby('anio').departamento_id.nunique()
```

```python
nacional = panel.groupby('anio')[['repitentes','sobreedad','salidos_sin_pase','matricula_grados_comparables','matricula_inicial']].sum(min_count=1)
nacional['proporcion_repitentes'] = nacional.repitentes / nacional.matricula_grados_comparables
nacional['proporcion_sobreedad'] = nacional.sobreedad / nacional.matricula_grados_comparables
nacional['proporcion_salidos_sin_pase'] = nacional.salidos_sin_pase / nacional.matricula_inicial
nacional.filter(like='proporcion_').plot(subplots=True, figsize=(9,8), title='Cocientes exploratorios nacionales')
```

```python
pd.crosstab(senales.indicador, senales.clasificacion)
```
